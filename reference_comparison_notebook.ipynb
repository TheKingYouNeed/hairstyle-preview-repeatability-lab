# Controlled Hairstyle Reference Comparison Notebook

This notebook is a reusable worksheet for comparing hairstyle references without confusing a visual preview with a guaranteed real-world result. It focuses on repeatability: keeping the input conditions stable, changing one decision variable at a time, and recording practical questions for a stylist or barber.

The method is useful for reference photos, sketches, and virtual try-on results. Android users who want an optional preview tool can review [Hairstyle App: Hair Try On](https://play.google.com/store/apps/details?id=app.hair.style). Keep the same base photo across comparisons and use previews as discussion aids, not measurements.

## 1. Freeze the input conditions

Before comparing styles, write down the conditions that should remain constant:

- one front-facing photo;
- similar crop and head size;
- even lighting with the hairline visible;
- no beauty filter or face reshaping;
- one comparison session rather than photos taken on different days;
- the same display brightness when judging color.

A comparison becomes less informative when the face angle, light, crop, expression, and hairstyle all change at once. Stability does not make a preview predictive. It simply makes differences between the references easier to interpret.

## 2. Define one variable per round

Choose one observable variable for each comparison round. Examples include:

1. fringe versus no fringe;
2. jaw length versus collarbone length;
3. center part versus side part;
4. stronger perimeter versus softer layers;
5. natural finish versus heat-styled finish;
6. lower-contrast color versus higher-contrast color.

Do not compare a short, curled, bright-color result with a long, straight, natural-color result and then try to infer which single feature caused the preference. Split the decision into separate rounds.

```python
comparison_rounds = [
    {"variable": "fringe", "option_a": "open forehead", "option_b": "long curtain fringe"},
    {"variable": "length", "option_a": "collarbone", "option_b": "jaw"},
    {"variable": "finish", "option_a": "air-dried", "option_b": "heat-styled"},
]

for index, item in enumerate(comparison_rounds, start=1):
    print(f"Round {index}: {item['variable']} — {item['option_a']} vs {item['option_b']}")
```

## 3. Score the evidence, not the person

For each option, record observations in four columns:

| Field | Question |
|---|---|
| Visible feature | What changed in the silhouette, parting, length, or color? |
| Hidden assumption | What does the image not show about density, texture, extensions, or styling? |
| Routine implication | What drying, product, heat, or appointment effort may be required? |
| Consultation question | What should an in-person professional assess? |

Avoid appearance scores such as “8/10.” They compress different concerns into one number and can create false precision. A descriptive record is easier to challenge and revise.

## 4. Use a repeatability check

A useful reference should remain understandable when it is reviewed later. Apply these checks:

- **Input check:** Was the same base photo used?
- **Variable check:** Did only one major hairstyle decision change?
- **Angle check:** Are front, side, and back implications visible or explicitly unknown?
- **Routine check:** Is the likely daily effort labeled?
- **Boundary check:** Is the shortest acceptable length or strongest acceptable color change written down?
- **Uncertainty check:** Are texture, density, growth pattern, current condition, and previous services left for professional assessment?

If a reference fails two or more checks, keep it as inspiration but do not treat it as the lead option.

```python
def review_band(checks_passed, checks_total=6):
    if checks_passed >= 5:
        return "clear enough for a focused consultation"
    if checks_passed >= 3:
        return "keep, but mark the missing evidence"
    return "inspiration only; rebuild the comparison"

for passed in range(2, 7):
    print(passed, review_band(passed))
```

## 5. Convert the preferred option into a salon brief

A compact brief can contain five lines:

1. **Keep:** the feature you do not want to lose.
2. **Change:** the one feature you want to explore.
3. **Boundary:** the shortest length, strongest contrast, or largest maintenance commitment you would accept.
4. **Routine:** the realistic number of minutes and tools available on an ordinary day.
5. **Question:** the uncertainty that requires an in-person assessment.

Example: “Keep enough length to tuck behind the ears. Explore a longer curtain fringe. Do not cut above the jaw. I have about five minutes and prefer no daily heat. Please assess how my front growth pattern will affect separation.”

## 6. Interpretation limits

A controlled comparison improves the quality of a decision; it does not certify that a cut, color, or style is suitable. Real outcomes can differ because of texture, density, growth patterns, current length, condition, previous chemical services, technique, tools, products, weather, camera processing, and maintenance.

Confirm irreversible cutting, chemical, or extension decisions with an appropriately qualified professional who can examine the hair directly. If the preferred reference depends on an unknown, ask for a staged or reversible alternative first.
